```python

# ============================================
# VANILLA PLANT DISEASE CNN - VERSION 2
# Transfer Learning (MobileNetV2) + Class Weighting
# Fixes: low recall on "black spots" class from v1
# Saves directly to Google Drive so nothing is lost on disconnect
# ============================================

# ============================================
# STEP 0: MOUNT GOOGLE DRIVE (run this first, always)
# ============================================
from google.colab import drive
drive.mount('/content/drive')

import os
PROJECT_DIR = "/content/drive/MyDrive/vanilla_cnn"
os.makedirs(PROJECT_DIR, exist_ok=True)
print("Project files will be saved to:", PROJECT_DIR)

# ============================================
# STEP 1: IMPORT LIBRARIES
# ============================================
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

print("====================================")
print("CNN V2 - VANILLA DISEASE (TRANSFER LEARNING)")
print("====================================")
print("TensorFlow version:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices('GPU'))

# ============================================
# STEP 2: DOWNLOAD KAGGLE DATASET
# ============================================
get_ipython().system('pip install -q kagglehub')
import kagglehub

dataset_path = kagglehub.dataset_download(
    "mihsanpermana/vanilla-plant-disease-image-dataset"
)
print("Dataset path:", dataset_path)

data_dir = os.path.join(dataset_path, "vanilla dataset")
classes = ["Healthy", "black spots", "red spots", "rotten stem"]
valid_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

# ============================================
# STEP 3: TRAIN / VALIDATION / TEST SPLIT
# (writing split to Drive this time, so it survives disconnects too —
#  comment out and point to existing split dir if you already have one saved)
# ============================================
import shutil
import random

random.seed(42)

split_dir = os.path.join(PROJECT_DIR, "vanilla_split")
train_dir = os.path.join(split_dir, "train")
val_dir = os.path.join(split_dir, "validation")
test_dir = os.path.join(split_dir, "test")

if not os.path.exists(split_dir):
    for directory in [train_dir, val_dir, test_dir]:
        for class_name in classes:
            os.makedirs(os.path.join(directory, class_name), exist_ok=True)

    for class_name in classes:
        source_dir = os.path.join(data_dir, class_name)
        images = [f for f in os.listdir(source_dir) if f.lower().endswith(valid_extensions)]
        random.shuffle(images)

        total = len(images)
        train_count = int(total * 0.70)
        val_count = int(total * 0.15)

        train_images = images[:train_count]
        val_images = images[train_count:train_count + val_count]
        test_images = images[train_count + val_count:]

        for image in train_images:
            shutil.copy2(os.path.join(source_dir, image), os.path.join(train_dir, class_name, image))
        for image in val_images:
            shutil.copy2(os.path.join(source_dir, image), os.path.join(val_dir, class_name, image))
        for image in test_images:
            shutil.copy2(os.path.join(source_dir, image), os.path.join(test_dir, class_name, image))

        print(f"{class_name}: Train={len(train_images)}, Val={len(val_images)}, Test={len(test_images)}")

    print("\nDataset split completed and saved to Drive.")
else:
    print("Split already exists in Drive — reusing it.")

# ============================================
# STEP 4: LOAD DATASETS
# ============================================
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

train_dataset = tf.keras.utils.image_dataset_from_directory(
    train_dir, labels="inferred", label_mode="int",
    image_size=IMG_SIZE, batch_size=BATCH_SIZE, shuffle=True, seed=SEED
)
validation_dataset = tf.keras.utils.image_dataset_from_directory(
    val_dir, labels="inferred", label_mode="int",
    image_size=IMG_SIZE, batch_size=BATCH_SIZE, shuffle=False
)
test_dataset = tf.keras.utils.image_dataset_from_directory(
    test_dir, labels="inferred", label_mode="int",
    image_size=IMG_SIZE, batch_size=BATCH_SIZE, shuffle=False
)

class_names = train_dataset.class_names
print("Classes:", class_names)

# ============================================
# STEP 5: COMPUTE CLASS WEIGHTS
# (this is the key fix for the black-spots recall problem —
#  it forces the loss function to penalize mistakes on weak classes more)
# ============================================
from sklearn.utils.class_weight import compute_class_weight

train_labels = []
for _, labels in train_dataset.unbatch():
    train_labels.append(labels.numpy())
train_labels = np.array(train_labels)

class_weights_array = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_labels),
    y=train_labels
)
class_weight_dict = dict(enumerate(class_weights_array))
print("Class weights:", class_weight_dict)

# ============================================
# STEP 6: NORMALIZATION + AUGMENTATION
# NOTE: MobileNetV2 expects inputs in [-1, 1], not [0, 1] —
# using its dedicated preprocess_input instead of a manual Rescaling layer
# ============================================
preprocess_input = keras.applications.mobilenet_v2.preprocess_input

data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.10),
    layers.RandomZoom(0.10),
    layers.RandomTranslation(height_factor=0.05, width_factor=0.05)
])

train_dataset_augmented = train_dataset.map(
    lambda x, y: (preprocess_input(data_augmentation(x)), y),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(tf.data.AUTOTUNE)

validation_dataset_processed = validation_dataset.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(tf.data.AUTOTUNE)

test_dataset_processed = test_dataset.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(tf.data.AUTOTUNE)

# ============================================
# STEP 7: BUILD MODEL - MOBILENETV2 TRANSFER LEARNING
# ============================================
base_model = keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)
base_model.trainable = False  # freeze for stage 1

model = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(len(classes), activation="softmax")
])

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-4),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# ============================================
# STEP 8: STAGE 1 TRAINING (frozen base)
# ============================================
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

checkpoint_path = os.path.join(PROJECT_DIR, "best_vanilla_cnn_v2.keras")

early_stopping = EarlyStopping(
    monitor="val_loss", patience=5, restore_best_weights=True, verbose=1
)
checkpoint = ModelCheckpoint(
    checkpoint_path, monitor="val_accuracy", save_best_only=True, mode="max", verbose=1
)

print("====================================")
print("STAGE 1: TRAINING WITH FROZEN BASE")
print("====================================")

history_stage1 = model.fit(
    train_dataset_augmented,
    validation_data=validation_dataset_processed,
    epochs=15,
    class_weight=class_weight_dict,
    callbacks=[early_stopping, checkpoint]
)

# ============================================
# STEP 9: STAGE 2 - FINE-TUNE TOP LAYERS OF MOBILENETV2
# (unfreeze last ~30 layers for a small accuracy boost)
# ============================================
base_model.trainable = True
fine_tune_at = len(base_model.layers) - 30
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),  # much lower LR for fine-tuning
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("====================================")
print("STAGE 2: FINE-TUNING TOP LAYERS")
print("====================================")

history_stage2 = model.fit(
    train_dataset_augmented,
    validation_data=validation_dataset_processed,
    epochs=10,
    class_weight=class_weight_dict,
    callbacks=[early_stopping, checkpoint]
)

# ============================================
# STEP 10: PLOT COMBINED TRAINING HISTORY
# ============================================
def combine_history(h1, h2, key):
    return h1.history[key] + h2.history[key]

acc = combine_history(history_stage1, history_stage2, "accuracy")
val_acc = combine_history(history_stage1, history_stage2, "val_accuracy")
loss = combine_history(history_stage1, history_stage2, "loss")
val_loss = combine_history(history_stage1, history_stage2, "val_loss")

plt.figure(figsize=(8, 5))
plt.plot(acc, label="Training Accuracy")
plt.plot(val_acc, label="Validation Accuracy")
plt.axvline(x=len(history_stage1.history["accuracy"]) - 1, color="gray", linestyle="--", label="Fine-tuning starts")
plt.title("CNN V2 Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.savefig(os.path.join(PROJECT_DIR, "training_accuracy.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(loss, label="Training Loss")
plt.plot(val_loss, label="Validation Loss")
plt.axvline(x=len(history_stage1.history["loss"]) - 1, color="gray", linestyle="--", label="Fine-tuning starts")
plt.title("CNN V2 Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.savefig(os.path.join(PROJECT_DIR, "training_loss.png"))
plt.show()

# ============================================
# STEP 11: LOAD BEST MODEL AND EVALUATE ON TEST SET
# ============================================
model = tf.keras.models.load_model(checkpoint_path)

test_loss, test_accuracy = model.evaluate(test_dataset_processed, verbose=1)
print("\n====================================")
print("FINAL TEST RESULT")
print("====================================")
print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")

# ============================================
# STEP 12: CONFUSION MATRIX & CLASSIFICATION REPORT
# ============================================
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay

y_true = []
y_pred = []
for images, labels in test_dataset_processed:
    predictions = model.predict(images, verbose=0)
    predicted_classes = np.argmax(predictions, axis=1)
    y_true.extend(labels.numpy())
    y_pred.extend(predicted_classes)

y_true = np.array(y_true)
y_pred = np.array(y_pred)

print("====================================")
print("CLASSIFICATION REPORT")
print("====================================")
print(classification_report(y_true, y_pred, target_names=classes, digits=4))

cm = confusion_matrix(y_true, y_pred)
print("CONFUSION MATRIX")
print(cm)

plt.figure(figsize=(8, 6))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=classes)
disp.plot(cmap="Blues", values_format="d")
plt.title("Vanilla Plant Disease CNN V2 - Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.savefig(os.path.join(PROJECT_DIR, "confusion_matrix.png"))
plt.show()

# ============================================
# STEP 13: CONVERT TO TFLITE FOR RASPBERRY PI DEPLOYMENT
# ============================================
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()

tflite_path = os.path.join(PROJECT_DIR, "vanilla_cnn_v2.tflite")
with open(tflite_path, "wb") as f:
    f.write(tflite_model)

print("====================================")
print("ALL FILES SAVED TO GOOGLE DRIVE")
print("====================================")
print("Keras model:", checkpoint_path)
print("TFLite model (for Pi):", tflite_path)
print("Plots:", PROJECT_DIR)
print("\nDownload 'vanilla_cnn_v2.tflite' from Drive and copy it to the Raspberry Pi.")
```

```python
# ============================================================
# VANILLA CNN V2 - LIVE CAMERA TEST (FIXED VERSION)
# Google Colab - ONE CELL
#
# CHANGES FROM ORIGINAL:
# 1. FIXED preprocessing bug: was using /255.0 (wrong range),
#    now uses mobilenet_v2.preprocess_input (matches training)
# 2. ADDED confidence threshold rejection: if the model's top
#    prediction confidence is below THRESHOLD, it reports
#    "NOT A VANILLA PLANT" instead of forcing a wrong class
# ============================================================

import tensorflow as tf
import numpy as np
import cv2
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode
from PIL import Image
import io

# ============================================================
# 1. MOUNT GOOGLE DRIVE
# ============================================================
from google.colab import drive
drive.mount('/content/drive')

# ============================================================
# 2. MODEL PATH
# ============================================================
MODEL_PATH = "/content/drive/MyDrive/vanilla_cnn/vanilla_cnn_v2.tflite"

# ============================================================
# 3. LOAD TFLITE MODEL
# ============================================================
interpreter = tf.lite.Interpreter(model_path=MODEL_PATH)
interpreter.allocate_tensors()

input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print("======================================")
print(" VANILLA CNN V2 CAMERA TEST (FIXED)")
print("======================================")
print("Model loaded successfully!")
print("Input shape :", input_details[0]["shape"])
print("Input type  :", input_details[0]["dtype"])
print("Output shape:", output_details[0]["shape"])

# ============================================================
# 4. CLASS NAMES
# ============================================================
class_names = [
    "Healthy",
    "black spots",
    "red spots",
    "rotten stem"
]

# ============================================================
# 4b. CONFIDENCE THRESHOLD FOR "NOT VANILLA" REJECTION
# ============================================================
# If the model's best guess is below this, treat the image as
# not a recognized vanilla plant / disease class.
# Tune between 50-60 based on your own testing.
CONFIDENCE_THRESHOLD = 80.0

# ============================================================
# 5. CAMERA FUNCTION
# ============================================================
def take_photo(quality=0.8):
    js = Javascript("""
    async function takePhoto(quality) {
        const div = document.createElement('div');
        const video = document.createElement('video');
        const stream =
            await navigator.mediaDevices.getUserMedia({
                video: true
            });

        document.body.appendChild(div);
        div.appendChild(video);
        video.srcObject = stream;
        await video.play();

        await new Promise(
            resolve => setTimeout(resolve, 3000)
        );

        const canvas =
            document.createElement('canvas');
        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;
        canvas.getContext('2d').drawImage(
            video,
            0,
            0
        );

        stream.getVideoTracks()[0].stop();
        div.remove();

        return canvas.toDataURL(
            'image/jpeg',
            quality
        );
    }
    """)
    display(js)
    data = eval_js(
        "takePhoto({})".format(quality)
    )
    binary = b64decode(
        data.split(",")[1]
    )
    image = Image.open(
        io.BytesIO(binary)
    )
    return np.array(image)

# ============================================================
# 6. TAKE PHOTO
# ============================================================
print("\nOpening camera...")
print("Allow camera permission when asked.")

image = take_photo()

# ============================================================
# 7. PREPROCESS IMAGE (FIXED)
# ============================================================
# Camera gives RGB image
image_rgb = image

# Resize to CNN input size
image_resized = cv2.resize(
    image_rgb,
    (224, 224)
)

# Convert to float32
image_resized = image_resized.astype(np.float32)

# ---- FIX: use MobileNetV2 preprocessing (range [-1, 1]) ----
# This MUST match what was used during training (Step 6 of the
# training notebook). The old script used `/255.0` which puts
# values in [0, 1] instead -- a mismatch that silently corrupts
# every prediction the model makes.
image_resized = tf.keras.applications.mobilenet_v2.preprocess_input(
    image_resized
)

# Add batch dimension
input_image = np.expand_dims(
    image_resized,
    axis=0
)

# ============================================================
# 8. CNN PREDICTION
# ============================================================
interpreter.set_tensor(
    input_details[0]["index"],
    input_image
)
interpreter.invoke()

output = interpreter.get_tensor(
    output_details[0]["index"]
)[0]

# ============================================================
# 9. CONVERT OUTPUT TO PROBABILITIES
# ============================================================
# If model output is already probabilities,
# use it directly. Otherwise apply softmax.
if np.sum(output) > 0.99 and np.sum(output) < 1.01:
    probabilities = output
else:
    probabilities = tf.nn.softmax(output).numpy()

# ============================================================
# 10. GET PREDICTION WITH "NOT VANILLA" REJECTION
# ============================================================
predicted_index = np.argmax(probabilities)
predicted_class = class_names[predicted_index]
confidence = probabilities[predicted_index] * 100

print("\n======================================")
print(" CNN PREDICTION RESULT")
print("======================================")

if confidence < CONFIDENCE_THRESHOLD:
    print("Result : NOT A VANILLA PLANT / UNRECOGNIZED IMAGE")
    print(
        f"(Highest class confidence was only {confidence:.2f}%, "
        f"below {CONFIDENCE_THRESHOLD}% threshold)"
    )
    final_decision = "NOT_VANILLA"
else:
    print("Predicted Class :", predicted_class)
    print("Confidence      :", f"{confidence:.2f}%")
    final_decision = predicted_class

print("\nClass Probabilities:")
print("--------------------------------------")
for name, probability in zip(class_names, probabilities):
    print(f"{name:15s} : {probability*100:.2f}%")
print("======================================")

# ============================================================
# 11. SHOW CAPTURED IMAGE
# ============================================================
display(
    Image.fromarray(image_rgb)
)

# ============================================================
# 12. FINAL DECISION (for robot logic / downstream use)
# ============================================================
print("\nfinal_decision variable =", final_decision)
```

```python
from google.colab import files
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.utils import load_img, img_to_array

# ============================================================
# CONFIDENCE THRESHOLD
# ============================================================

THRESHOLD = 80.0   # Change to 90.0 if you want 90%

# ============================================================
# UPLOAD IMAGE
# ============================================================

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

# ============================================================
# LOAD IMAGE
# ============================================================

img = load_img(
    image_path,
    target_size=(224, 224)
)

img_array = img_to_array(img)

img_array = np.expand_dims(
    img_array,
    axis=0
)

# ============================================================
# MOBILENETV2 PREPROCESSING
# ============================================================

img_array = tf.keras.applications.mobilenet_v2.preprocess_input(
    img_array
)

# ============================================================
# PREDICTION
# ============================================================

prediction = model.predict(
    img_array,
    verbose=0
)

# ============================================================
# GET PREDICTION
# ============================================================

predicted_class = np.argmax(prediction[0])

confidence = prediction[0][predicted_class] * 100

# ============================================================
# APPLY THRESHOLD
# ============================================================

if confidence >= THRESHOLD:

    final_prediction = class_names[predicted_class]
    status = "ACCEPTED"

else:

    final_prediction = "UNKNOWN / NOT CONFIDENT"
    status = "REJECTED"

# ============================================================
# DISPLAY RESULT
# ============================================================

print("\n==============================")
print("       CNN PREDICTION")
print("==============================")

print("Image       :", image_path)
print("Prediction  :", final_prediction)
print("Confidence  :", f"{confidence:.2f}%")
print("Threshold   :", f"{THRESHOLD:.0f}%")
print("Status      :", status)

# ============================================================
# CLASS PROBABILITIES
# ============================================================

print("\nClass probabilities:")
print("------------------------------")

for name, prob in zip(class_names, prediction[0]):

    print(
        f"{name:15s}: {prob*100:.2f}%"
    )

# ============================================================
# SPRAY DECISION
# ============================================================

print("\n==============================")
print("       SPRAY DECISION")
print("==============================")

if final_prediction == "UNKNOWN / NOT CONFIDENT":

    print("❌ UNKNOWN / NOT CONFIDENT")
    print("❌ DO NOT SPRAY")

elif final_prediction == "Healthy":

    print("🌿 HEALTHY LEAF")
    print("❌ DO NOT SPRAY")

else:

    print("⚠️ DISEASE DETECTED")
    print("Disease :", final_prediction)
    print("✅ SPRAY CAN BE ACTIVATED")

print("==============================")

# ============================================================
# DISPLAY IMAGE
# ============================================================

plt.figure(figsize=(6, 6))

plt.imshow(img)
plt.axis("off")

plt.title(
    f"{final_prediction} ({confidence:.2f}%)"
)

plt.show()
```
